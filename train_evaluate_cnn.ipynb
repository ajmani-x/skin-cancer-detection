```python
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
```

```python
IMG_SIZE = 128
BATCH_SIZE = 30
EPOCHS = 60
```

```python
TRAIN_DIR = "Dataset/data/train"
TEST_DIR = "Dataset/data/test"
```

```python
train_datagen = ImageDataGenerator(
    rescale=1./255,
    shear_range=0.1,
    zoom_range=0.2  ,
    horizontal_flip=True
)
```

```python
test_datagen = ImageDataGenerator(
    rescale=1./255
)
```

```python
train_generator = train_datagen.flow_from_directory(
    TRAIN_DIR,
    target_size=(128, 128),
    batch_size=BATCH_SIZE,
    class_mode="binary"
)
```

```python
test_generator = test_datagen.flow_from_directory(
    TEST_DIR,
    target_size=(128, 128),
    batch_size=BATCH_SIZE,
    class_mode="binary",
    shuffle=False
)
```

```python
model = Sequential()
```

```python
model.add(Conv2D(filters=15, kernel_size=(3,3), activation='relu', input_shape=(128, 128, 3)))
model.add(MaxPooling2D(2,2))
```

```python
model.add(Conv2D(filters=10, kernel_size=(3,3), activation='relu'))
model.add(MaxPooling2D(2,2))
```

```python
model.add(Conv2D(7, (3,3), activation='relu'))
model.add(MaxPooling2D(2,2))
```

```python
model.add(Flatten())
```

```python
model.add(Dense(16, activation='relu'))
```

```python
model.add(Dropout(0.2))
```

```python
model.add(Dense(1, activation='sigmoid')) 
```

```python
model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.summary()
```

```python
history = model.fit(
    train_generator,
    steps_per_epoch=88,
    epochs=EPOCHS,
    validation_data=test_generator,
)
```

```python
model.save("cnn_malignant_vs_benign_v2.h5")
```

```python
test_generator.reset()
```

```python
pred_probs = model.predict(test_generator)
pred_labels = (pred_probs > 0.5).astype(int)
```

```python
true_labels = test_generator.classes
```

```python
test_loss, test_acc = model.evaluate(test_generator)
print(f"\n✅ Test Accuracy: {test_acc * 100:.2f}%")
```

```python
print("\n📊 Classification Report:")
print(classification_report(true_labels, pred_labels, target_names=["Benign", "Malignant"]))
```
